```python
# Remove impossible ages
df = df[df['age'] > 0]

# Fill missing MonthlyIncome with median
df['MonthlyIncome'] = df['MonthlyIncome'].fillna(
    df['MonthlyIncome'].median()
)

# Fill missing dependents
df['NumberOfDependents'] = df['NumberOfDependents'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df['IncomePerDependent'] = (
    df['MonthlyIncome'] /
    (df['NumberOfDependents'] + 1)
)
```

```python
df.isnull().sum()
```

```python
df[['MonthlyIncome','NumberOfDependents','IncomePerDependent']].head()
```

```python
df = df.drop('Unnamed: 0', axis=1)
```

```python
df.columns
```

```python
X = df.drop('SeriousDlqin2yrs', axis=1)
y = df['SeriousDlqin2yrs']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'
)

lr.fit(X_train_scaled, y_train)
```

```python
preds = lr.predict(X_test_scaled)

probs = lr.predict_proba(X_test_scaled)[:, 1]
```

```python
from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    confusion_matrix
)

print(classification_report(y_test, preds))
print("ROC-AUC:", roc_auc_score(y_test, probs))
print(confusion_matrix(y_test, preds))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
rf_preds = rf.predict(X_test)

rf_probs = rf.predict_proba(X_test)[:,1]
```

```python
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

print(classification_report(y_test, rf_preds))

print("ROC-AUC:", roc_auc_score(y_test, rf_probs))

print(confusion_matrix(y_test, rf_preds))
```
